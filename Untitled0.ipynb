```python
df = pd.read_parquet('/content/train-00000-of-00001.parquet', engine="pyarrow")
```

```python
def translate_to_egyptian(text):
    fusha = translator(text, src_lang="eng_Latn", tgt_lang="arb_Arab", max_length=1000)[0]['translation_text']
    prompt = f"حوّل هذه الجملة من العربية الفصحى إلى اللهجة المصرية: {fusha}"
    egy = translator(prompt, src_lang="arb_Arab", tgt_lang="arb_Arab", max_length=1000)[0]['translation_text']
    return egy

translated = []
for p in df['text']:
    try:
        translated.append(translate_to_egyptian(p))
    except Exception as e:
        print("Error:", e)
        translated.append(p)

df["egyptian_prompt"] = translated
df.to_parquet("translated_egyptian.parquet", index=False)

from google.colab import files
files.download("translated_egyptian.parquet")
```
